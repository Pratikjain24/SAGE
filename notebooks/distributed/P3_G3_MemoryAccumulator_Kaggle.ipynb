```python
# Step 9: Provenance & Authenticity Audit
# Scans all produced trajectory events to mathematically verify zero mock fallback.
%cd /kaggle/working/SAGE
import json, glob

trajectories = glob.glob('/kaggle/working/SAGE/experiments/runs/study_live_G3_*/trajectory.jsonl')
total_tasks = real_calls = zero_tok = fallback_calls = 0
total_in = total_out = 0
models_seen = set()
seeds_seen = set()
seen_starts = set()
dup_starts = 0
timeouts_by_seed = {}
phantom_successes = 0

for traj_path in trajectories:
    for line in open(traj_path, 'r', encoding='utf-8'):
        line = line.strip()
        if not line: continue
        ev = json.loads(line)
        etype = ev.get('event_type')
        s = ev.get('seed')
        if etype == 'task_start':
            k = (s, ev.get('cycle'), ev.get('task_id'))
            if k in seen_starts: dup_starts += 1
            seen_starts.add(k)
        elif etype == 'task_end':
            total_tasks += 1
            p = ev.get('payload', {})
            c = ev.get('cost', {})
            tin = int(c.get('tokens_in', 0) or 0)
            tout = int(c.get('tokens_out', 0) or 0)
            total_in += tin
            total_out += tout
            if p.get('is_fallback'): fallback_calls += 1
            if (tin + tout) == 0: zero_tok += 1
            else: real_calls += 1
            if p.get('model_name'): models_seen.add(p['model_name'])
            if s is not None:
                seeds_seen.add(s)
                if s not in timeouts_by_seed: timeouts_by_seed[s] = 0
                if p.get('status') == 'timeout' or ev.get('status') == 'timeout':
                    timeouts_by_seed[s] += 1
            if p.get('success') and (p.get('ground_truth_score', 0) < 0.5 or (tin + tout) == 0):
                phantom_successes += 1

print('=' * 65)
print('📊 PROVENANCE & AUTHENTICITY AUDIT REPORT — Layer G3')
print('=' * 65)
print(f'Total Evaluated Tasks : {total_tasks}')
print(f'Real LLM Inferences   : {real_calls}')
print(f'Zero-Token Invocations: {zero_tok}')
print(f'Mock Fallbacks        : {fallback_calls}')
print(f'Duplicate Task Starts : {dup_starts}')
print(f'Phantom Successes     : {phantom_successes}')
print(f'Total Prompt Tokens   : {total_in:,}')
print(f'Total Gen Tokens      : {total_out:,}')
print(f'Audited Seeds         : {sorted(seeds_seen)}')
print(f'Timeouts by Seed      : {timeouts_by_seed}')
print(f'Recorded Models       : {sorted(models_seen)}')
print('=' * 65)
excess_timeouts = {sd: cnt for sd, cnt in timeouts_by_seed.items() if cnt > 5}
if fallback_calls == 0 and zero_tok == 0 and real_calls > 0 and dup_starts == 0 and len(excess_timeouts) == 0 and phantom_successes == 0:
    print('✅ SCIENTIFIC AUDIT PASSED: 100% Genuine Empirical Inferences Recorded!')
    print('   - 0 Mock fallbacks, 0 zero-token tasks, 0 duplicate starts, <= 5 timeouts/seed.')
else:
    print('⚠️ WARNING: Review trajectory logs before submitting:')
    if dup_starts > 0: print(f'   ❌ {dup_starts} duplicate task start events found.')
    if excess_timeouts: print(f'   ❌ Invalid seeds with >5 timeouts: {excess_timeouts}')
    if fallback_calls > 0: print(f'   ❌ {fallback_calls} mock fallbacks detected.')
    if zero_tok > 0: print(f'   ❌ {zero_tok} zero-token tasks detected.')
    if phantom_successes > 0: print(f'   ❌ {phantom_successes} phantom successes detected.')
```

```python
# Step 10: Generate Publication LaTeX Tables for Layer G3
# Generates table1_main_results_live.tex directly from empirical results.
%cd /kaggle/working/SAGE
import glob, json, pathlib
import numpy as np

td = pathlib.Path('/kaggle/working/SAGE/paper/tables')
td.mkdir(parents=True, exist_ok=True)

metrics_files = sorted(glob.glob('/kaggle/working/SAGE/experiments/runs/study_live_G3_*/results/cycle_metrics.json'))
all_metrics = []
for mf in metrics_files:
    try:
        all_metrics.extend(json.loads(pathlib.Path(mf).read_text(encoding='utf-8')))
    except Exception:
        pass

if not all_metrics:
    print('No metrics available to render LaTeX tables yet.')
else:
    seeds = sorted(set(m.get('seed', 42) for m in all_metrics))
    cycs = sorted(set(m.get('cycle', 0) for m in all_metrics))
    c0 = cycs[0]
    c_tr_last = [c for c in cycs if c < 4][-1] if any(c < 4 for c in cycs) else c0
    cT = cycs[-1]
    p_tr0 = float(np.mean([m['success_rate'] for m in all_metrics if m['cycle'] == c0]))
    p_tr_last = float(np.mean([m['success_rate'] for m in all_metrics if m['cycle'] == c_tr_last]))
    d_tr = p_tr_last - p_tr0
    p_test = float(np.mean([m['success_rate'] for m in all_metrics if m['cycle'] == cT]))
    dr = float(np.mean([m.get('safety_drift', 0.0) for m in all_metrics if m['cycle'] == cT]))
    co = float(sum(m.get('cost_usd', 0.0) for m in all_metrics))
    d_tr_s = f'+{d_tr:.3f}' if d_tr >= 0 else f'{d_tr:.3f}'
    dr_s   = f'+{dr:.3f}'   if dr > 0   else f'{dr:.3f}'
    seed_str = ', '.join(str(s) for s in seeds)

    row = f'G3 & Memory Accumulator & {p_tr0:.3f} & {p_tr_last:.3f} & {d_tr_s} & {p_test:.3f} & {dr_s} & \\${co:.4f} \\\\'

    header = [
        '\\begin{table*}[t]',
        '\\centering',
        '\\small',
        '% Auto-generated for Layer G3 (Memory Accumulator) — 100% empirical Kaggle Dual T4 data',
        '\\caption{\\textbf{Empirical Benchmark Results for Layer G3}. Seeds: ({seed_str}), N={len(seeds)} seeds, T={cT+1} cycles. $P_{\\text{train}}$ on 60 train tasks; $P_{\\text{test}}$ on 40 held-out tasks; $\\Delta P_{\\text{train}}$ measures true training adaptation.}',
        '\\label{tab:live_g3}',
        '\\begin{tabular}{llcccccc}',
        '\\toprule',
        r'Group & Mechanism & $P_{\text{train}}(0)$ & $P_{\text{train}}(3)$ & $\Delta P_{\text{train}}$ & $P_{\text{test}}(4)$ & SafetyDrift & API Cost(\$) \\',
        '\\midrule',
    ]
    footer = ['\\bottomrule', '\\end{tabular}', '\\end{table*}']
    out1 = td / 'table1_g3_results_live.tex'
    out1.write_text('\n'.join(header + [row] + footer), encoding='utf-8')
    print(f'✓ LaTeX Table 1 written: {out1}\n')
    print(out1.read_text(encoding='utf-8'))
```

```python
# Step 11: Render Publication Figures for Layer G3
# Generates publication-ready PNG plots with matplotlib.
%cd /kaggle/working/SAGE
import glob, json, pathlib, matplotlib
matplotlib.use('Agg')
import matplotlib.pyplot as plt
import numpy as np
from IPython.display import Image, display

figs_dir = pathlib.Path('/kaggle/working/SAGE/paper/figures')
figs_dir.mkdir(parents=True, exist_ok=True)

metrics_files = sorted(glob.glob('/kaggle/working/SAGE/experiments/runs/study_live_G3_*/results/cycle_metrics.json'))
all_metrics = []
for mf in metrics_files:
    try:
        all_metrics.extend(json.loads(pathlib.Path(mf).read_text(encoding='utf-8')))
    except Exception:
        pass

if not all_metrics:
    print('No metrics available to render figures.')
else:
    cycs = sorted(set(m.get('cycle', 0) for m in all_metrics))
    seeds = sorted(set(m.get('seed', 42) for m in all_metrics))
    color_map = {
        'G1': '#64748b',
        'G2': '#3b82f6',
        'G3': '#10b981',
        'G4': '#f59e0b',
        'G5': '#8b5cf6',
        'G6': '#ec4899',
        'G7': '#ef4444',
    }
    c_val = color_map.get('G3', '#2563eb')

    for mkey, ylabel, title, fname in [
        ('success_rate', 'Success Rate P(t)', 'Capability Progression — G3: Memory Accumulator', 'capability_g3_live.png'),
        ('safety_drift', 'Safety Drift V(t) - V(0)', 'Safety Boundary Drift — G3: Memory Accumulator', 'safety_drift_g3_live.png'),
    ]:
        fig, ax = plt.subplots(figsize=(7.5, 4.5), dpi=150)
        mean_vals = [float(np.mean([m[mkey] for m in all_metrics if m['cycle'] == c])) for c in cycs]
        std_vals = [float(np.std([m[mkey] for m in all_metrics if m['cycle'] == c])) if len([m for m in all_metrics if m['cycle'] == c]) > 1 else 0.0 for c in cycs]
        
        ax.plot(cycs, mean_vals, marker='o', label=f'G3: Memory Accumulator', color=c_val, linewidth=2.4, markersize=6)
        if any(s > 0 for s in std_vals):
            ax.fill_between(cycs, np.array(mean_vals) - np.array(std_vals), np.array(mean_vals) + np.array(std_vals), color=c_val, alpha=0.18, label='±1 Std Dev')
        
        ax.set_xlabel('Evolutionary Cycle (t)', fontsize=11, fontweight='bold')
        ax.set_ylabel(ylabel, fontsize=11, fontweight='bold')
        ax.set_title(f'{title} (N={len(seeds)} Seeds)', fontsize=12, fontweight='bold')
        ax.legend(frameon=True, fontsize=9, loc='best')
        ax.grid(True, linestyle='--', alpha=0.5)
        fig.tight_layout()
        
        out_fig = figs_dir / fname
        fig.savefig(out_fig)
        plt.close(fig)
        print(f'✓ Generated Figure: {out_fig}')
        display(Image(str(out_fig)))
```

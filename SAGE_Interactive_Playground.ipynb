### Overview
**SAGE (Safety & Agent Growth Evaluator)** is a hardened benchmark and formal verification framework designed to measure **security boundary drift (vulnerability injection rate), specification gaming (proxy gaps), and rollback-guarded capability retention** in self-modifying code agents.

**Key Highlights of this Interactive Playground:**
- 🚀 **Zero-Setup & 100% Free**: Enabled by default! No OpenAI, Anthropic, or external API keys needed. Runs on standard free Colab CPU or GPU in <60 seconds.
- 🔬 **Full 7-Archetype Benchmark Matrix**: Simulates G1 (Frozen Control), G2 (Prompt Rewriter), G3 (Memory Accumulator), G4 (Reflection Agent), G5 (Static Verifier Gate), G6 (Regression-Guarded Verifier), and G7 (Deployable Proxy Canary Guard).
- 📊 **Live Visualizations**: Plots Security Boundary Drift vs. Cycles, Capability Trajectories, Specification Gaming, and Rollback Retention.
- 🔐 **SAGE-Live Attestation**: Demonstrates deterministic grammar probe generation, AST invariant checking, and Ed25519/HMAC-SHA256 cryptographic block sealing.
- ⚖️ **Policy-as-Code Compliance**: Auto-generates EU AI Act Annex III and ISO/IEC 42001 compliance draft reports.
- 🛡️ **Stateful Multi-Turn Red-Teaming**: Simulates multi-turn privilege escalation intercepted by canary defenses.
- 🎛️ **Optional Live API Toggle**: Easily test live frontier models (Claude 3.5 Sonnet, GPT-4o, DeepSeek, Gemini, Qwen) whenever you provide your key.

## Step 5 — SAGE-Live Dynamic Evaluation Workbench & Cryptographic Attestation
> Generates grammar algebra dynamic probes, evaluates agent code against active AST invariants, and seals Ed25519 / HMAC-SHA256 blocks.

````python
# Step 5: SAGE-Live Dynamic Evaluation & Attestation Ledger
from sage_live.core.probe_generator import ProbeGenerator
from sage_live.core.attestation import AttestationService, KeyPair
import hashlib, uuid, time

print('=' * 70)
print('SAGE-LIVE RUNTIME ENGINE: DYNAMIC PROBES & IMMUTABLE LEDGER')
print('=' * 70)

# 1. Generate Deterministic Probes via Grammar Algebra
gen = ProbeGenerator(generation=1)
probes = gen.generate_batch(count=3)
print(f'\n[1] Dynamic Probes Generated (Count={len(probes)}):')
for idx, p in enumerate(probes, 1):
    print(f'   {idx}. ID: {p.probe_id} | Vuln: {p.vulnerability_class} | Difficulty: {p.difficulty:.3f}')

# 2. Initialize Attestation Ledger
kp = KeyPair()
attest = AttestationService(key_pair=kp, db_path='experiments/attestation_playground.db')

# 3. Seal Tamper-Evident Ledger Block
sample_result = {
    'evaluation_id': str(uuid.uuid4()),
    'agent_id': 'G7-DeployableCanary',
    'probe_id': probes[0].probe_id,
    'composite_safety_score': 88.5,
    'safety_drift_score': 1.8,
    'proxy_gap_score': 0.015,
    'passed': True,
}
block = attest.record_and_seal(results=[sample_result])
print(f'\n[2] Cryptographic Attestation Block Sealed:')
print(f'   - Window ID     : {block.window_id}')
print(f'   - Current Hash  : {block.current_hash[:24]}...')
print(f'   - Previous Hash : {block.previous_hash[:24]}...')
print(f'   - Signature     : {block.signature[:32]}... (Ed25519)')

# 4. Cryptographic Hash Chain Verification
valid, msg = attest.verify_chain_integrity()
print(f'\n[3] Ledger Cryptographic Verification: {"PASS (VALID)" if valid else "FAIL"} | {msg}')
print('=' * 70)
````
